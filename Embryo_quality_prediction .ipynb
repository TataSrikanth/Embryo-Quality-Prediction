```python
from ultralytics import YOLO

# Load a model
model = YOLO("yolov8x-cls.pt")  # load a pretrained model (recommended for training)

# Train the model
results = model.train(data="/content/Final_data/Final_data2", epochs=20, imgsz=352)
```

```python
# prompt: predict and display image for this path /content/data/skin_type_project/test/oily/cropped_chin_1696929695_jpg.rf.1cad3de7ad873e49218fcb4e6b99d30c.jpg

from PIL import Image
import matplotlib.pyplot as plt
from ultralytics import YOLO

model = YOLO("/content/runs/classify/train/weights/best.pt")

# Path to the image
image_path = r"/content/Final_data/Final_data2/test/Blastocyst_Grade_A/Blastocyst_A (10)_0_8067.jpeg"
# Perform inference
results = model(image_path)

# Display the image with predictions
for r in results:
    im_array = r.plot()  # plot a BGR numpy array of predictions
    im = Image.fromarray(im_array[..., ::-1])  # RGB PIL image
    plt.imshow(im)
    plt.show()
```

```python
names = results[0].names
names[results[0].probs.top1]
```

```python
probs = results[0].probs  # Probabilities for each class
predicted_labels = results[0].names  # Class names (labels)

# Find the index of the highest probability
max_prob_idx = probs.top1  # Index of highest probabilit
# Get the label and probability corresponding to the highest confidence score
highest_confidence_label = predicted_labels[max_prob_idx]
highest_confidence_score = probs.top1conf.item()
```
